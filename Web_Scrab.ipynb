```python
import pandas as pd
import csv
import requests
from bs4 import BeautifulSoup

Books_details = []

def main_code():
  for page_num in range(1 , 6):
    url = f'https://books.toscrape.com/catalogue/page-{page_num}.html'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    books = soup.find_all('li', class_='col-xs-6 col-sm-4 col-md-3 col-lg-3')

    for index , book in enumerate(books , 1) :
      # id => index
      book_name = book.find('h3').find('a')['title']
      book_rate = book.find('p', class_ = 'star-rating')['class'][1]
      book_price = book.find('p', class_ = 'price_color').text
      book_link = book.find('a')['href']
      # book_auther =
      # publishdate =

      Books_details.append({
      'Id' : index ,
      'Name' : book_name ,
      'Rate_Stars' : book_rate ,
      'Price'  : book_price ,
      'Link ' : book_link
      #    'Auther'  : book_auther
      #    'publishdate' : publishdate

      })

def csv_printing():
  header = Books_details[0].keys()
  path = 'Book_Fair_collected.csv'
  with open(path,'w', newline ='') as File :
    writer = csv.DictWriter(File , header)
    writer.writeheader()
    writer.writerows(Books_details)
    print ('File made successfully')

main_code()
csv_printing()
```

# Step 3 - cleaning

```python
# view values of rate column
print(df['Rate_Stars'].unique())
```

```python
# Handle Rate format - transform rate to integers
integers = {
    'One' : 1 ,
    'Two' : 2 ,
    'Three' : 3 ,
    'Four' : 4 ,
    'Five' : 5
}
df['Rate_Stars']= df['Rate_Stars'].replace(integers)
```

```python
# Delete Â£ from prices column & transform price to numbers
df['Price']= df['Price'].str.replace("Â£","").astype(float)
```
